* **題目說明**：將帶有符號與空白的原始文字資料進行清理與型別轉換，依序計算 10% 服務費、扣除折價券後的應付總額、每人分攤費用，並透過布林比較判斷是否超過每人 300 元的預算。

```python
# === 測試一：原始資料 ===
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"
service_rate = 0.10
discount = 100
budget_per_person = 300

clean_name = event_name.strip().lower()
meal_cost = float(meal_amount.replace("NT$", "").replace(",", "").strip())
people_count = int(people)

service_fee = meal_cost * service_rate
total_amount = meal_cost + service_fee - discount
cost_per_person = total_amount / people_count
is_over_budget = cost_per_person > budget_per_person

print("【測試一】")
print("聚餐名稱：", clean_name)
print("原始餐費：", meal_cost)
print("服務費：", service_fee)
print("折價券：", discount)
print("應付總額：", total_amount)
print("分攤人數：", people_count)
print("每人費用：", cost_per_person)
print("每人預算：", budget_per_person)
print("是否超過預算：", is_over_budget)
```
